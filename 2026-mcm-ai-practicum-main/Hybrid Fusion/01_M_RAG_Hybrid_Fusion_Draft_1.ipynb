```python
import os
import json
import fitz
from PIL import Image
from ultralytics import YOLO

# ─── CONFIG ───────────────────────────────────────────────
INPUT_FILE    = "/content/mrag-research-paper"
OUTPUT_DIR    = "output/images"
JSON_FILE     = "output/image_text.json"
MODEL_PATH    = "/content/yolov8s-doclaynet.pt"
DPI           = 200
CONFIDENCE    = 0.5
VICINITY      = 50    # PDF units — how far "nearby" means
# ──────────────────────────────────────────────────────────

FIGURE_LABELS = {"picture", "figure"}

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs("output", exist_ok=True)

model = YOLO(MODEL_PATH)

doc   = fitz.open(INPUT_FILE)
zoom  = DPI / 72
mat   = fitz.Matrix(zoom, zoom)
total = 0

result_json = {}  # { image_path: nearby_text }

for page_num in range(len(doc)):
    page = doc[page_num]
    pix  = page.get_pixmap(matrix=mat)
    page_image = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)

    results = model(page_image, conf=CONFIDENCE)[0]

    fig_count = 0
    for box, cls, conf in zip(results.boxes.xyxy, results.boxes.cls, results.boxes.conf):
        label = model.names[int(cls)].lower()
        if label not in FIGURE_LABELS:
            continue

        # ── Save image ──────────────────────────────────────
        x1, y1, x2, y2 = [int(c) for c in box]
        cropped  = page_image.crop((x1, y1, x2, y2))
        filename = f"page{page_num + 1}_figure{fig_count + 1}.png"
        img_path = os.path.join(OUTPUT_DIR, filename)
        cropped.save(img_path)

        # ── Get nearby text ──────────────────────────────────
        # Convert pixel bbox back to PDF units for text search
        pdf_x1 = x1 / zoom
        pdf_y1 = y1 / zoom
        pdf_x2 = x2 / zoom
        pdf_y2 = y2 / zoom

        # Expand the box by VICINITY to capture surrounding text
        search_rect = fitz.Rect(
            pdf_x1 - VICINITY,
            pdf_y1 - VICINITY,
            pdf_x2 + VICINITY,
            pdf_y2 + VICINITY
        )

        # Extract text only within that expanded region
        nearby_text = page.get_text("text", clip=search_rect).strip()

        # ── Store in JSON ────────────────────────────────────
        result_json[img_path] = nearby_text

        print(f"  Saved : {filename}  (conf: {conf:.2f})")
        print(f"  Text  : {nearby_text[:80]}{'...' if len(nearby_text) > 80 else ''}")

        fig_count += 1
        total += 1

    print(f"Page {page_num + 1}: {fig_count} figure(s) found\n")

doc.close()

# ── Write JSON ───────────────────────────────────────────
with open(JSON_FILE, "w", encoding="utf-8") as f:
    json.dump(result_json, f, indent=4, ensure_ascii=False)

print(f"[✓] Done! {total} images saved to : {OUTPUT_DIR}/")
print(f"[✓] JSON saved to                 : {JSON_FILE}")
```

```python
import json
import base64
import io
from pathlib import Path
from PIL import Image


# ---------------- PATHS ----------------
METADATA_PATH = Path("output/image_metadata.json")


# ---------------- BUILD PAGE TEXT MAP FROM DOCLING ----------------
page_text_map = {}

for element, level in doc.iterate_items():

    if hasattr(element, "prov") and element.prov:
        page_no = element.prov[0].page_no
        text = getattr(element, "text", "")

        if text:
            page_text_map.setdefault(page_no, []).append(text)

page_text_map = {
    page_no: "\n".join(texts)
    for page_no, texts in page_text_map.items()
}

print(f"Built text map for {len(page_text_map)} pages")


# ---------------- IMAGE TO BASE64 ----------------
def image_to_base64(image_path, max_size=1024, quality=70):
    image = Image.open(image_path)

    if image.mode != "RGB":
        image = image.convert("RGB")

    image.thumbnail((max_size, max_size))

    buffer = io.BytesIO()
    image.save(buffer, format="JPEG", quality=quality, optimize=True)

    return base64.b64encode(buffer.getvalue()).decode("utf-8")


# ---------------- LOAD IMAGE METADATA ----------------
with open(METADATA_PATH, "r", encoding="utf-8") as f:
    image_metadata = json.load(f)


# ---------------- ADD DESCRIPTIONS USING PAGE TEXT ----------------
for image_id, metadata in image_metadata.items():

    image_path = metadata["image_path"]
    page_no = metadata.get("page_no")

    nearby_text = page_text_map.get(page_no, "")

    try:
        image_base64 = image_to_base64(image_path)

        description = generate_description(
            image_base64=image_base64,
            nearby_text=nearby_text
        )

        metadata["nearby_text"] = nearby_text[:1500]
        metadata["description"] = description.strip()

        print(f"Processed image {image_id} from page {page_no}")

    except Exception as e:
        metadata["description"] = ""
        metadata["error"] = str(e)

        print(f"Failed image {image_id} from page {page_no}: {e}")


# ---------------- SAVE UPDATED METADATA ----------------
with open(METADATA_PATH, "w", encoding="utf-8") as f:
    json.dump(image_metadata, f, ensure_ascii=False, indent=2)

print("Updated image_metadata.json with nearby_text and descriptions")
```

```python
import json
import numpy as np
import networkx as nx
import community.community_louvain as community_louvain

from pathlib import Path
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ---------------- PATHS ----------------
CHUNKS_PATH = Path("output/chunks.json")
IMAGES_PATH = Path("output/image_metadata.json")

EARLY_FUSION_PATH = Path("output/early_fusion.json")
LATE_FUSION_PATH = Path("output/late_fusion.json")


# ---------------- LOAD DATA ----------------
with open(CHUNKS_PATH, "r", encoding="utf-8") as f:
    text_chunks = json.load(f)

with open(IMAGES_PATH, "r", encoding="utf-8") as f:
    image_metadata = json.load(f)


# ---------------- PREPARE TEXT + IMAGE DESCRIPTIONS ----------------
text_ids = list(text_chunks.keys())
text_values = list(text_chunks.values())

image_ids = []
image_descriptions = []

for image_id, meta in image_metadata.items():
    desc = meta.get("description", "").strip()

    if desc:
        image_ids.append(image_id)
        image_descriptions.append(desc)


print("Text chunks:", len(text_ids))
print("Images with descriptions:", len(image_ids))


# ---------------- EMBEDDINGS ----------------
model = SentenceTransformer("all-MiniLM-L6-v2")

text_embeddings = model.encode(
    text_values,
    normalize_embeddings=True,
    show_progress_bar=True
)

image_desc_embeddings = model.encode(
    image_descriptions,
    normalize_embeddings=True,
    show_progress_bar=True
)


# ---------------- COSINE SIMILARITY ----------------
similarity_matrix = cosine_similarity(
    text_embeddings,
    image_desc_embeddings
)


# ---------------- BUILD BIPARTITE GRAPH ----------------
G = nx.Graph()

# Add text nodes
for text_id in text_ids:
    G.add_node(f"text::{text_id}", node_type="text", original_id=text_id)

# Add image nodes
for image_id in image_ids:
    G.add_node(f"image::{image_id}", node_type="image", original_id=image_id)


# Add weighted edges
SIMILARITY_THRESHOLD = 0.25   # tune this
TOP_K_IMAGES_PER_TEXT = 3    # prevents noisy full graph

for i, text_id in enumerate(text_ids):

    top_indices = np.argsort(similarity_matrix[i])[::-1][:TOP_K_IMAGES_PER_TEXT]

    for j in top_indices:
        score = float(similarity_matrix[i][j])

        if score >= SIMILARITY_THRESHOLD:
            image_id = image_ids[j]

            G.add_edge(
                f"text::{text_id}",
                f"image::{image_id}",
                weight=score
            )


print("Graph nodes:", G.number_of_nodes())
print("Graph edges:", G.number_of_edges())


# ---------------- LOUVAIN CLUSTERING ----------------
partition = community_louvain.best_partition(
    G,
    weight="weight",
    random_state=42
)

# node -> cluster_id
node_to_cluster = partition


# ---------------- GROUP NODES BY CLUSTER ----------------
clusters = {}

for node, cluster_id in node_to_cluster.items():
    clusters.setdefault(cluster_id, {
        "text_ids": [],
        "image_ids": []
    })

    node_type = G.nodes[node]["node_type"]
    original_id = G.nodes[node]["original_id"]

    if node_type == "text":
        clusters[cluster_id]["text_ids"].append(original_id)
    else:
        clusters[cluster_id]["image_ids"].append(original_id)


# ---------------- EARLY FUSION ----------------
early_fusion = {}

for cluster_id, items in clusters.items():

    cluster_text_ids = items["text_ids"]
    cluster_image_ids = items["image_ids"]

    # only early fuse if cluster has both text and image
    if not cluster_text_ids or not cluster_image_ids:
        continue

    fused_parts = []

    for tid in cluster_text_ids:
        fused_parts.append(text_chunks[tid])

    for iid in cluster_image_ids:
        fused_parts.append(image_metadata[iid]["description"])

    fusion_key = tuple(cluster_text_ids + cluster_image_ids)

    early_fusion[str(fusion_key)] = {
        "cluster_id": cluster_id,
        "text_ids": cluster_text_ids,
        "image_ids": cluster_image_ids,
        "fused_text": "\n\n".join(fused_parts)
    }


# ---------------- LATE FUSION ----------------
late_fusion = {}

for i, text_id in enumerate(text_ids):
    text_cluster = node_to_cluster.get(f"text::{text_id}")

    for j, image_id in enumerate(image_ids):
        image_cluster = node_to_cluster.get(f"image::{image_id}")

        if text_cluster != image_cluster:
            score = float(similarity_matrix[i][j])

            key = str((text_id, image_id))

            late_fusion[key] = {
                "text_id": text_id,
                "image_id": image_id,
                "text": text_chunks[text_id],
                "image_description": image_metadata[image_id]["description"],
                "similarity_score": score,
                "text_cluster": text_cluster,
                "image_cluster": image_cluster
            }


# ---------------- SAVE JSON FILES ----------------
with open(EARLY_FUSION_PATH, "w", encoding="utf-8") as f:
    json.dump(early_fusion, f, ensure_ascii=False, indent=2)

with open(LATE_FUSION_PATH, "w", encoding="utf-8") as f:
    json.dump(late_fusion, f, ensure_ascii=False, indent=2)


print(f"Saved early fusion JSON: {EARLY_FUSION_PATH}")
print(f"Saved late fusion JSON: {LATE_FUSION_PATH}")

print("Early fusion groups:", len(early_fusion))
print("Late fusion pairs:", len(late_fusion))
```

```python
import json
from pathlib import Path

import torch
import chromadb
import numpy as np
from PIL import Image
from tqdm import tqdm

from transformers import CLIPProcessor, CLIPModel
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import Ridge


# ---------------- PATHS ----------------
LATE_FUSION_PATH = Path("output/late_fusion.json")
IMAGE_METADATA_PATH = Path("output/image_metadata.json")
CHROMA_DIR = "output/chroma_db"


# ---------------- LOAD DATA ----------------
with open(LATE_FUSION_PATH, "r", encoding="utf-8") as f:
    late_fusion = json.load(f)

with open(IMAGE_METADATA_PATH, "r", encoding="utf-8") as f:
    image_metadata = json.load(f)


# ---------------- MODELS ----------------
device = "cuda" if torch.cuda.is_available() else "cpu"

text_model = SentenceTransformer("all-MiniLM-L6-v2")

clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

clip_model.eval()


# ---------------- HELPERS ----------------
def normalize_vector(vector):
    norm = np.linalg.norm(vector)
    if norm == 0:
        return vector
    return vector / norm


def get_text_embedding(text):
    embedding = text_model.encode(
        text,
        normalize_embeddings=True
    )
    return np.array(embedding)


def get_image_embedding(image_path):
    image = Image.open(image_path).convert("RGB")

    inputs = clip_processor(
        images=image,
        return_tensors="pt"
    )

    pixel_values = inputs["pixel_values"].to(device)

    with torch.no_grad():
        outputs = clip_model.vision_model(pixel_values=pixel_values)
        image_features = outputs.pooler_output

        image_features = clip_model.visual_projection(image_features)

    image_features = image_features.detach().cpu().numpy()[0]
    image_features = normalize_vector(image_features)

    return image_features


# ---------------- PREPARE TRAINING PAIRS ----------------
X_image = []
Y_text = []
valid_items = []

for pair_key, item in tqdm(late_fusion.items(), desc="Preparing pairs"):

    text_id = item["text_id"]
    image_id = item["image_id"]

    text = item["text"]

    if image_id not in image_metadata:
        continue

    image_path = image_metadata[image_id]["image_path"]

    try:
        text_emb = get_text_embedding(text)
        image_emb = get_image_embedding(image_path)

        X_image.append(image_emb)
        Y_text.append(text_emb)

        valid_items.append((pair_key, item))

    except Exception as e:
        print(f"Skipped {pair_key}: {e}")


X_image = np.array(X_image)
Y_text = np.array(Y_text)

print("Valid pairs:", len(valid_items))
print("Image embedding shape:", X_image.shape)
print("Text embedding shape:", Y_text.shape)


# ---------------- TRAIN LINEAR MAPPING ----------------
# CLIP image space -> text embedding space

linear_mapper = Ridge(alpha=1.0)

linear_mapper.fit(X_image, Y_text)

print("Linear mapping trained")
print("Image dim:", X_image.shape[1])
print("Text dim:", Y_text.shape[1])


# ---------------- VECTOR DATABASE ----------------
client = chromadb.PersistentClient(path=CHROMA_DIR)

collection = client.get_or_create_collection(
    name="late_fusion_collection"
)


# ---------------- STORE LATE FUSION VECTORS ----------------
for pair_key, item in tqdm(valid_items, desc="Storing late fusion vectors"):

    text_id = item["text_id"]
    image_id = item["image_id"]

    text = item["text"]
    image_path = image_metadata[image_id]["image_path"]

    try:
        text_emb = get_text_embedding(text)
        image_emb = get_image_embedding(image_path)

        # Map image vector into text vector space
        mapped_image_emb = linear_mapper.predict([image_emb])[0]
        mapped_image_emb = normalize_vector(mapped_image_emb)

        # Late fusion: average text vector + mapped image vector
        fused_emb = (text_emb + mapped_image_emb) / 2
        fused_emb = normalize_vector(fused_emb)

        vector_id = f"late_{text_id[:16]}_{image_id[:16]}"

        metadata = {
            "fusion_type": "late",
            "text_id": text_id,
            "image_id": image_id,
            "tuple_ids": json.dumps((text_id, image_id)),
            "image_path": image_path,
            "similarity_score": float(item.get("similarity_score", 0.0)),
            "text_cluster": item.get("text_cluster"),
            "image_cluster": item.get("image_cluster")
        }

        document = text + "\n\n[IMAGE]\n" + image_path

        collection.add(
            ids=[vector_id],
            embeddings=[fused_emb.tolist()],
            documents=[document],
            metadatas=[metadata]
        )

    except Exception as e:
        print(f"Failed {pair_key}: {e}")


print("Total late fusion vectors stored:", collection.count())
```

```python
client = chromadb.PersistentClient(path=CHROMA_DIR)

image_collection = client.get_or_create_collection(
    name="image_only_collection"
)

stored = 0

for image_id, metadata in image_metadata.items():

    image_path = metadata["image_path"]

    try:
        # Original CLIP embedding (512)
        image_emb = get_image_embedding(image_path)

        # Map into text space (384)
        mapped_image_emb = linear_mapper.predict([image_emb])[0]

        mapped_image_emb = normalize_vector(mapped_image_emb)

        description = metadata.get("description", "")

        image_collection.add(
            ids=[image_id],
            embeddings=[mapped_image_emb.tolist()],
            documents=[description],
            metadatas=[{
                "image_id": image_id,
                "image_path": image_path,
                "page_no": metadata.get("page_no"),
                "fusion_type": "image_only"
            }]
        )

        stored += 1

    except Exception as e:
        print(f"Failed {image_id}: {e}")

print(f"Stored {stored} image vectors")
print(f"Collection count: {image_collection.count()}")
```

```python
import re
import numpy as np
import networkx as nx

from sentence_transformers import SentenceTransformer, util


# =====================================================
# INPUT
# =====================================================
# answer
# top_5
# image_metadata
# linear_mapper
# get_image_embedding()
# =====================================================


# ---------------- MODEL ----------------
model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)


# ---------------- SPLIT ANSWER ----------------
sentences = [
    s.strip()
    for s in re.split(
        r'(?<=[.!?])\s+',
        answer
    )
    if s.strip()
]


# ---------------- EXTRACT UNIQUE IMAGES ----------------
candidate_images = {}

for item in top_5:

    metadata = item.get("metadata", {})

    if not isinstance(metadata, dict):
        continue

    image_id = metadata.get("image_id")

    if image_id is None:
        continue

    if image_id not in image_metadata:
        continue

    candidate_images[image_id] = image_metadata[
        image_id
    ]


print(
    "Candidate Images:",
    len(candidate_images)
)


# ---------------- SENTENCE EMBEDDINGS ----------------
sentence_embeddings = model.encode(
    sentences,
    convert_to_tensor=True,
    normalize_embeddings=True
)


# ---------------- IMAGE EMBEDDINGS ----------------
image_ids = []
mapped_image_embeddings = []

for image_id, image_info in candidate_images.items():

    image_path = image_info["image_path"]

    try:

        clip_embedding = get_image_embedding(
            image_path
        )

        mapped_embedding = linear_mapper.predict(
            [clip_embedding]
        )[0]

        mapped_embedding = (
            mapped_embedding
            /
            np.linalg.norm(mapped_embedding)
        )

        image_ids.append(
            image_id
        )

        mapped_image_embeddings.append(
            mapped_embedding
        )

    except Exception as e:

        print(
            f"Failed {image_id}: {e}"
        )


mapped_image_embeddings = np.array(
    mapped_image_embeddings
)

print(
    "Mapped Images:",
    len(image_ids)
)


# ---------------- BUILD GRAPH ----------------
G = nx.Graph()

SIMILARITY_THRESHOLD = 0.30


# Sentence nodes
for sentence_idx in range(
    len(sentences)
):

    G.add_node(
        f"s_{sentence_idx}",
        bipartite=0
    )


# Image nodes
for image_idx in range(
    len(image_ids)
):

    G.add_node(
        f"img_{image_idx}",
        bipartite=1
    )


# ---------------- ADD WEIGHTED EDGES ----------------
for sentence_idx in range(
    len(sentences)
):

    sentence_embedding = (
        sentence_embeddings[
            sentence_idx
        ]
        .cpu()
        .numpy()
    )

    for image_idx in range(
        len(image_ids)
    ):

        image_embedding = (
            mapped_image_embeddings[
                image_idx
            ]
        )

        similarity = util.cos_sim(
            sentence_embedding,
            image_embedding
        ).item()

        if similarity >= SIMILARITY_THRESHOLD:

            G.add_edge(
                f"s_{sentence_idx}",
                f"img_{image_idx}",
                weight=similarity
            )


print(
    "Sentence Nodes:",
    len(sentences)
)

print(
    "Image Nodes:",
    len(image_ids)
)

print(
    "Edges:",
    G.number_of_edges()
)


# ---------------- BLOSSOM MATCHING ----------------
matching = nx.algorithms.matching.max_weight_matching(
    G,
    maxcardinality=True,
    weight="weight"
)


# ---------------- SENTENCE -> IMAGE ----------------
sentence_to_image = {}

for node_a, node_b in matching:

    if node_a.startswith("s_"):

        sentence_idx = int(
            node_a.split("_")[1]
        )

        image_idx = int(
            node_b.split("_")[1]
        )

    else:

        sentence_idx = int(
            node_b.split("_")[1]
        )

        image_idx = int(
            node_a.split("_")[1]
        )

    sentence_to_image[
        sentence_idx
    ] = image_ids[
        image_idx
    ]


print(
    "Matched Images:",
    len(sentence_to_image)
)


# ---------------- BUILD MULTIMODAL ANSWER ----------------
multimodal_answer = []

for sentence_idx, sentence in enumerate(
    sentences
):

    multimodal_answer.append({
        "type": "text",
        "content": sentence
    })

    if sentence_idx in sentence_to_image:

        image_id = sentence_to_image[
            sentence_idx
        ]

        multimodal_answer.append({
            "type": "image",
            "image_id": image_id,
            "image_path": candidate_images[
                image_id
            ]["image_path"]
        })


# ---------------- DISPLAY ----------------
for item in multimodal_answer:

    if item["type"] == "text":

        print(item["content"])

    else:

        print(
            f"\n[INSERT IMAGE: "
            f"{item['image_id']}]"
        )

        print(
            item["image_path"]
        )

        print()


# ---------------- OUTPUTS ----------------
#
# multimodal_answer
#
# sentence_to_image
#
# candidate_images
#
```
